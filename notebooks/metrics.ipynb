```python
import mlflow
import polars as pl
```

```python
exp = mlflow.get_experiment_by_name("etth1s")
runs = pl.from_dataframe(mlflow.search_runs(experiment_ids=[exp.experiment_id])).filter(
    status="FINISHED"
)
```

```python
runs.group_by("tags.method").agg(
    pl.format(
        "{} ({})",
        pl.col("metrics.test_crps").mean().round(4),
        pl.col("metrics.test_crps").std().round(4),
    ),
    pl.format(
        "{} ({})",
        pl.col("metrics.test_nmae").mean().round(4),
        pl.col("metrics.test_nmae").std().round(4),
    ),
    pl.format(
        "{} ({})",
        pl.col("metrics.test_mse").mean().round(4),
        pl.col("metrics.test_mse").std().round(4),
    ),
)
```
